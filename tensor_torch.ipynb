```python
import torch
import numpy as np
```

```python
data=[[1, 2], [3, 4]]
x_data=torch.tensor(data)
```

```python
print(type(x_data))
```

```python
np_array = np.array(data)
```

```python
print(type(np_array))
```

```python
numpy_to_tensor = torch.from_numpy(np_array)
print(type(numpy_to_tensor))
```

```python
print(numpy_to_tensor)
```

```python
x_ones=torch.ones_like(x_data)
```

```python
print(x_ones)
```

```python
x_rand = torch.rand_like(x_data,dtype=torch.float)
print(x_rand)
```

```python
shape = (2, 3,)
tensor=torch.tensor([[0.8858, 0.8944],
        [0.7361, 0.6226]])
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

```python
print(torch.tensor(tensor))
```

```python
# We move our tensor to the GPU if available
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else 'cpu'
tensor = tensor.to("mps " if device == "mps" else device)
print(f"Device tensor is stored on: {tensor.device}")
```

```python
tensor = torch.ones(4, 4)
tensor[:, 1] = 0
print(tensor)
```

```python
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

```python
# This computes the element-wise product
print(f"tensor.mul(tensor) \n {tensor.mul(5)} \n")
# Alternative syntax:
print(f"tensor * tensor \n {tensor * tensor}")
```

```python
print(tensor.T)
```

```python
print(f"tensor.matmul(tensor.T) \n {tensor.matmul(tensor.T)} \n")
# Alternative syntax:
print(f"tensor @ tensor.T \n {tensor @ tensor.T}")
```
